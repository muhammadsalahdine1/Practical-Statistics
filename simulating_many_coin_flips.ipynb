```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# number of heads from 10 fair coin flips   >>>> n = 10 , p = 0.5
np.random.binomial(10,0.5)
```

```python
# results from 20 tests with 10 coin flips  >>>> n = 10 , p = 0.5 , size = 20
np.random.binomial(10,0.5,size=20)
```

```python
# mean number of heads from the 20 tests
np.random.binomial(10,0.5,size=20).mean()
```

```python
# reflects the fairness of the coin more closely as # tests increases
np.random.binomial(10,0.5,size=1000000).mean()
```

```python
plt.hist(np.random.binomial(10,0.5,size=1000000));
```

```python
# gets more narrow as number of flips increase per test
plt.hist(np.random.binomial(100, 0.5, 1000000));
```
